```python
from pyspark.sql.types import StructType , StructField , IntegerType , DoubleType , StringType
schema = StructType([
    StructField("id ", IntegerType(), True),
    StructField("movie_id", StringType(), True),
    StructField("rating_1", IntegerType(), True),
    StructField("timestamp", DoubleType(), True)
])
df=spark.read.format("csv")\
    .option("header", "true")\
        .schema(schema)\
            .load("/Volumes/bigdata2/my_voloum/volume/ratings.csv")
```

```python
from pyspark.sql import functions as f

df.withColumn("newid", f.col("id") + 1)\
    .withColumn("newrating", f.col("rating_1")*10)\
        .select("movie_id", "newid", "newrating").show()
```

```python
from pyspark.sql import functions as f
from pyspark.sql.types import *

df = spark.read.csv(
    "/Volumes/bigdata2/my_voloum/volume/ratings.csv",
    header=True,
    inferSchema=True
)

df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import to_timestamp,from_unixtime

df = df.withColumn(
    "unique_timestamp",
    to_timestamp(from_unixtime(f.col("unique_timestamp").cast("long")))
)

df.printSchema()
```
