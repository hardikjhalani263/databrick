```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

spark = spark.builder.appName("BasicWindow").getOrCreate()

data = [
    ("HR","Emp1",3000), ("HR","Emp2",3200), ("HR","Emp3",2800),
    ("HR","Emp4",3500), ("HR","Emp5",3100), ("HR","Emp6",3300),
    ("HR","Emp7",3400), ("HR","Emp8",3600), ("HR","Emp9",2900),
    ("HR","Emp10",3000),

    ("IT","Emp11",5000), ("IT","Emp12",5200), ("IT","Emp13",4800),
    ("IT","Emp14",5500), ("IT","Emp15",5100), ("IT","Emp16",5300),
    ("IT","Emp17",5400), ("IT","Emp18",5600), ("IT","Emp19",4900),
    ("IT","Emp20",5000),

    ("Sales","Emp21",4000), ("Sales","Emp22",4200), ("Sales","Emp23",3900),
    ("Sales","Emp24",4500), ("Sales","Emp25",4100), ("Sales","Emp26",4300),
    ("Sales","Emp27",4400), ("Sales","Emp28",4600), ("Sales","Emp29",3950),
    ("Sales","Emp30",4050),

    ("Finance","Emp31",6000), ("Finance","Emp32",6200), ("Finance","Emp33",5800),
    ("Finance","Emp34",6500), ("Finance","Emp35",6100), ("Finance","Emp36",6300),
    ("Finance","Emp37",6400), ("Finance","Emp38",6600), ("Finance","Emp39",5900),
    ("Finance","Emp40",6050),

    ("Admin","Emp41",2500), ("Admin","Emp42",2600), ("Admin","Emp43",2400),
    ("Admin","Emp44",2700), ("Admin","Emp45",2550), ("Admin","Emp46",2650),
    ("Admin","Emp47",2750), ("Admin","Emp48",2800), ("Admin","Emp49",2450),
    ("Admin","Emp50",2500)
]

columns = ["department","employee","salary"]

df = spark.createDataFrame(data, columns)

df.show(50, False)

# COMMAND ----------

windowSpec = Window.partitionBy("department")

# COMMAND ----------

df.withColumn(
    "dept_avg_salary",
    avg("salary").over(windowSpec)
).show()
```

```python
windowSpec = Window.partitionBy("department").orderBy("salary")

df.withColumn(
    "dept_total_salary",
    sum("salary").over(windowSpec) ).show()
```

```python
# assigns sequence row number inside each department.
windowSpec = Window.partitionBy("department").orderBy("salary")

df.withColumn(
    "row_num",
    row_number().over(windowSpec)
).show()
```

```python
# same value get same rank (with gaps)
df.withColumn("rank", rank().over(windowSpec)).show()
```

```python
# running total (cummlative sum)
windowSpec = Window.partitionBy("department").orderBy("salary").rowsBetween(Window.unboundedPreceding,
                                                                            Window.currentRow)
df.withColumn(
    "running_total",
    sum("salary").over(windowSpec)
).show()
```

```python
# Average salary per department 
WindowSepc = Window.partitionBy("department")

df.withColumn("dept_avg_salary", avg("salary").over(WindowSepc)).show()
```

```python
# Lag (previous ROW value)
windowSpec = Window.partitionBy("department").orderBy("salary")

df.withColumn(
    "prev_salary",
    lag("salary", 1).over(windowSpec)
).show()
```

```python
# top earner per department
windowSpec = Window.partitionBy("department").orderBy(col("salary").desc())

df.withColumn("rank" , row_number().over(windowSpec))\
    .filter(col("rank") == 1)\
    .show()
    
```

```python
df_with_avg = df.withColumn(
    "dept_avg_salary",
    avg("salary").over(windowSpec)
)
```

```python
df_with_diff = df_with_avg.withColumn(
    "salary_diff",
    round(col("salary") - col("dept_avg_salary"), 2)
)
```

```python
final_df = df_with_diff.withColumn(
    "salary_vs_dept_avg",
    when(col("salary_diff") > 0, "Greater than Avg")
    .when(col("salary_diff") < 0, "Less than Avg")
    .otherwise("Equal to Avg")
)
```

```python
final_df.select(
    "department",
    "employee",
    "salary",
    round("dept_avg_salary", 2).alias("dept_avg_salary"),
    "salary_diff",
    "salary_vs_dept_avg"
).show(50, False)
```
