```python
from pyspark.sql import functions as f

base_url_s3 = "s3://jhalani-bucket/city/original_data/*.json"

city_df = spark.read.option("multiline", "true").json(base_url_s3) \
    .withColumn("load_timestamp", f.current_timestamp()) \
    .select("*", "_metadata.file_name") # Fixed typo: _metadata

display(city_df)
```
